## Leakage Hunt

I checked that none of the selected features directly contain the prediction target or future information. Fields such as `trend_direction` and `trend_pct` are excluded because they describe the outcome being predicted. Only features available before the prediction time are included.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
leaky_columns = [
    "trend_direction",
    "trend_pct"
]

print("Potential leakage columns:")
for col in leaky_columns:
    if col in df.columns:
        print("-", col)
```
